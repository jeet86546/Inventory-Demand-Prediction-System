```python
import pandas as pd
import numpy as np
import warnings
import joblib
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.preprocessing import LabelEncoder

warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("walmart_cleaned.csv", parse_dates=["Date"])

print("Dataset Shape:", df.shape)

df.head()
```

```python
#Column names
df.columns
```

```python
#Dataset information
df.info()
```

```python
#Statistical summary
df.describe()
```

```python
#Check Missing Values
df.isnull().sum()
```

```python
#Handle Missing Values
df = df.dropna()
```

```python
# Time-based features
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Week"] = df["Date"].dt.isocalendar().week.astype(int)
df["Quarter"] = df["Date"].dt.quarter

# Convert holiday boolean → integer
df["IsHoliday"] = df["IsHoliday"].astype(int)

df.head()
```

```python
# Check missing values
print(df.isnull().sum())

# Fill missing values
df.fillna(0, inplace=True)

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
# Encode Store Type (A/B/C)

type_encoder = LabelEncoder()

df["Type"] = type_encoder.fit_transform(df["Type"])

df[["Type"]].head()
```

```python
#Distribution of Weekly Sales (Histogram)

plt.figure(figsize=(8,5))
plt.hist(df['Weekly_Sales'], bins=50)
plt.title("Distribution of Weekly Sales")
plt.xlabel("Weekly Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
#Average Sales by Store (Bar Chart)

store_sales = df.groupby('Store')['Weekly_Sales'].mean()
plt.figure(figsize=(10,5))
store_sales.plot(kind='bar')
plt.title("Average Weekly Sales by Store")
plt.xlabel("Store")
plt.ylabel("Average Sales")
plt.show()
```

```python
#Average Sales by Department

dept_sales = df.groupby('Dept')['Weekly_Sales'].mean()
plt.figure(figsize=(12,5))
dept_sales.plot(kind='bar')
plt.title("Average Weekly Sales by Department")
plt.xlabel("Department")
plt.ylabel("Average Sales")
plt.show()
```

```python
#Holiday vs Non-Holiday Sales

holiday_sales = df.groupby('IsHoliday')['Weekly_Sales'].mean()
plt.figure(figsize=(6,4))
holiday_sales.plot(kind='bar')
plt.title("Holiday vs Non-Holiday Sales")
plt.xlabel("Is Holiday")
plt.ylabel("Average Weekly Sales")
plt.show()
```

```python
#Year-wise Sales Trend (Line Plot)

year_sales = df.groupby('Year')['Weekly_Sales'].mean()
plt.figure(figsize=(8,5))
plt.plot(year_sales.index, year_sales.values)
plt.title("Year-wise Average Sales Trend")
plt.xlabel("Year")
plt.ylabel("Average Weekly Sales")
plt.show()
```

```python
#Holiday vs Non-Holiday Sales Contribution
#(Shows how much sales come from holidays vs normal days)

holiday_total = df.groupby('IsHoliday')['Weekly_Sales'].sum()

plt.figure(figsize=(6,6))
plt.pie(
    holiday_total,
    labels=['Non-Holiday', 'Holiday'],
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Sales Contribution: Holiday vs Non-Holiday")
plt.show()
```

```python
#Sales Contribution by Year
#(Good for showing long-term trend)

year_total_sales = df.groupby('Year')['Weekly_Sales'].sum()

plt.figure(figsize=(7,7))
plt.pie(
    year_total_sales,
    labels=year_total_sales.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Year-wise Sales Contribution")
plt.show()
```

```python
#Top 5 Stores Sales Contribution
#(Do NOT show all stores — pie chart becomes messy)

top_stores = (
    df.groupby('Store')['Weekly_Sales']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

plt.figure(figsize=(7,7))
plt.pie(
    top_stores,
    labels=top_stores.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Top 5 Stores Sales Contribution")
plt.show()
```

## Outlier Detection

```python
#Detect Outliers using Box Plot

plt.figure(figsize=(8,4))
plt.boxplot(df['Weekly_Sales'], vert=False)
plt.title("Box Plot of Weekly Sales")
plt.xlabel("Weekly Sales")
plt.show()
```

```python
#IQR Method(Calculate Quartiles)

Q1 = df['Weekly_Sales'].quantile(0.25)
Q3 = df['Weekly_Sales'].quantile(0.75)

IQR = Q3 - Q1
```

```python
#Define Lower & Upper Bound

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

lower_bound, upper_bound
```

```python
#Count Outliers

outliers = df[
    (df['Weekly_Sales'] < lower_bound) | 
    (df['Weekly_Sales'] > upper_bound)
]

outliers.shape

```

```python
#Remove Outliers

df = df[
    (df['Weekly_Sales'] >= lower_bound) & 
    (df['Weekly_Sales'] <= upper_bound)
]
```

```python
#Verify After Removal

plt.figure(figsize=(8,4))
plt.boxplot(df['Weekly_Sales'], vert=False)
plt.title("Box Plot After Outlier Removal")
plt.xlabel("Weekly Sales")
plt.show()
```

```python
FEATURES = [
    "Store",
    "Dept",
    "IsHoliday",
    "Year",
    "Month",
    "Week",
    "Quarter",
    "Temperature",
    "Fuel_Price",
    "CPI",
    "Unemployment",
    "Size",
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5",
    "Type"
]

TARGET = "Weekly_Sales"

X = df[FEATURES]
y = df[TARGET]

print("Features Shape:", X.shape)
```

```python
# Sort by date

df = df.sort_values("Date")

X = df[FEATURES]
y = df[TARGET]

# Time-based split
split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Train Shape:", X_train.shape)
print("Test Shape:", X_test.shape)
```

```python
rf_model = RandomForestRegressor(
    n_estimators=200,
    max_depth=20,
    min_samples_leaf=3,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

print("✅ Random Forest Training Completed")
```

```python
xgb_model = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=10,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

xgb_model.fit(X_train, y_train)

print("✅ XGBoost Training Completed")
```

```python
xgb_model = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=10,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

xgb_model.fit(X_train, y_train)

print("✅ XGBoost Training Completed")
```

```python
rf_pred = rf_model.predict(X_test)

xgb_pred = xgb_model.predict(X_test)
```

```python
def evaluate_model(y_true, y_pred, model_name):

    mae = mean_absolute_error(y_true, y_pred)

    rmse = np.sqrt(mean_squared_error(y_true, y_pred))

    r2 = r2_score(y_true, y_pred)

    mape = np.mean(
        np.abs((y_true - y_pred) / (y_true + 1e-6))
    ) * 100

    print(f"\n📊 {model_name} Performance")
    print("-" * 40)

    print("MAE :", round(mae, 2))
    print("RMSE:", round(rmse, 2))
    print("R²  :", round(r2, 4))
    print("MAPE:", round(mape, 2), "%")

    return mae, rmse, r2, mape
```

```python
rf_metrics = evaluate_model(
    y_test,
    rf_pred,
    "Random Forest"
)

xgb_metrics = evaluate_model(
    y_test,
    xgb_pred,
    "XGBoost"
)
```

```python
importance_df = pd.DataFrame({
    "Feature": FEATURES,
    "Importance": rf_model.feature_importances_
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
)

importance_df
```

```python
joblib.dump(rf_model, "model.pkl")

joblib.dump(xgb_model, "XGBoost.pkl")

joblib.dump(FEATURES, "features.pkl")

joblib.dump(type_encoder, "type_encoder.pkl")

print("✅ Models Saved Successfully")
```
