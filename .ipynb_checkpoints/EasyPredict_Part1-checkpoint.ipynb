## Project Name : Easy Predict - Sales Forecasting Using Machine Learning

Project Information:

• This project predicts future sales using historical Walmart sales data.

• Machine learning is used to analyze trends and improve inventory planning.

• The model helps distributors estimate future demand and reduce stock cost.

• Proper data preprocessing and visualization are used for accurate prediction.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load Dataset
df = pd.read_csv("train.csv")
```

```python
#View first 5 rows
df.head()
```

```python
#Shape of dataset
df.shape
```

```python
#Column names
df.columns
```

```python
#Dataset information
df.info()
```

```python
#Statistical summary
df.describe()
```

```python
#Check Missing Values
df.isnull().sum()
```

```python
#Handle Missing Values
df = df.dropna()
```

```python
#Convert Date to datetime
df['Date'] = pd.to_datetime(df['Date'])
```

```python
#Extract useful parts from date
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Week'] = df['Date'].dt.isocalendar().week
```

```python
#Check updated dataset
df.head()
```

```python
#Sort Data
df = df.sort_values(by='Date')
```

```python
#Remove Unnecessary Columns
df = df.drop(columns=['Date'])
```

## EDA

```python
#Distribution of Weekly Sales (Histogram)

plt.figure(figsize=(8,5))
plt.hist(df['Weekly_Sales'], bins=50)
plt.title("Distribution of Weekly Sales")
plt.xlabel("Weekly Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
#Average Sales by Store (Bar Chart)

store_sales = df.groupby('Store')['Weekly_Sales'].mean()
plt.figure(figsize=(10,5))
store_sales.plot(kind='bar')
plt.title("Average Weekly Sales by Store")
plt.xlabel("Store")
plt.ylabel("Average Sales")
plt.show()
```

```python
#Average Sales by Department

dept_sales = df.groupby('Dept')['Weekly_Sales'].mean()
plt.figure(figsize=(12,5))
dept_sales.plot(kind='bar')
plt.title("Average Weekly Sales by Department")
plt.xlabel("Department")
plt.ylabel("Average Sales")
plt.show()
```

```python
#Holiday vs Non-Holiday Sales

holiday_sales = df.groupby('IsHoliday')['Weekly_Sales'].mean()
plt.figure(figsize=(6,4))
holiday_sales.plot(kind='bar')
plt.title("Holiday vs Non-Holiday Sales")
plt.xlabel("Is Holiday")
plt.ylabel("Average Weekly Sales")
plt.show()
```

```python
#Year-wise Sales Trend (Line Plot)

year_sales = df.groupby('Year')['Weekly_Sales'].mean()
plt.figure(figsize=(8,5))
plt.plot(year_sales.index, year_sales.values)
plt.title("Year-wise Average Sales Trend")
plt.xlabel("Year")
plt.ylabel("Average Weekly Sales")
plt.show()
```

```python
#Holiday vs Non-Holiday Sales Contribution
#(Shows how much sales come from holidays vs normal days)

holiday_total = df.groupby('IsHoliday')['Weekly_Sales'].sum()

plt.figure(figsize=(6,6))
plt.pie(
    holiday_total,
    labels=['Non-Holiday', 'Holiday'],
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Sales Contribution: Holiday vs Non-Holiday")
plt.show()
```

```python
#Sales Contribution by Year
#(Good for showing long-term trend)

year_total_sales = df.groupby('Year')['Weekly_Sales'].sum()

plt.figure(figsize=(7,7))
plt.pie(
    year_total_sales,
    labels=year_total_sales.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Year-wise Sales Contribution")
plt.show()
```

```python
#Top 5 Stores Sales Contribution
#(Do NOT show all stores — pie chart becomes messy)

top_stores = (
    df.groupby('Store')['Weekly_Sales']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

plt.figure(figsize=(7,7))
plt.pie(
    top_stores,
    labels=top_stores.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Top 5 Stores Sales Contribution")
plt.show()
```

## Outlier Detection

```python
#Detect Outliers using Box Plot

plt.figure(figsize=(8,4))
plt.boxplot(df['Weekly_Sales'], vert=False)
plt.title("Box Plot of Weekly Sales")
plt.xlabel("Weekly Sales")
plt.show()
```

```python
#IQR Method(Calculate Quartiles)

Q1 = df['Weekly_Sales'].quantile(0.25)
Q3 = df['Weekly_Sales'].quantile(0.75)

IQR = Q3 - Q1
```

```python
#Define Lower & Upper Bound

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

lower_bound, upper_bound
```

```python
#Count Outliers

outliers = df[
    (df['Weekly_Sales'] < lower_bound) | 
    (df['Weekly_Sales'] > upper_bound)
]

outliers.shape

```

```python
#Remove Outliers

df = df[
    (df['Weekly_Sales'] >= lower_bound) & 
    (df['Weekly_Sales'] <= upper_bound)
]
```

```python
#Verify After Removal

plt.figure(figsize=(8,4))
plt.boxplot(df['Weekly_Sales'], vert=False)
plt.title("Box Plot After Outlier Removal")
plt.xlabel("Weekly Sales")
plt.show()
```
